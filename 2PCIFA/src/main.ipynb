```python
import os
import pickle
from pathlib import Path
from utils.statistic import *
from core.io import load_and_align_data
from core.calculate_overall_dff import calculate_t2p_dff
from core.dff_analysis import *
fig_save_format = 'png'
%matplotlib inline
%config InlineBackend.figure_format = fig_save_format
current_dir = Path(os.getcwd())
```

```python
# Parameters initiate
plot_pre = 10  # seconds for plot window before the event
plot_post = 10  # seconds for plot window after the event
pre_duration=2  # seconds for detecting cell activity before the event
post_duration=2  # seconds for detecting cell activity after the event
time_window = 1  # seconds for detecting cell activity in time window

groups_files_dir = current_dir.parent / "groups"
data_save_dir = current_dir.parent / "data"
fig_save_path = data_save_dir / "figures"
os.makedirs(fig_save_path, exist_ok=True)
track2p_datas_dir = data_save_dir / "data_t2p"
# Parameters for running bout analysis
running_parameters = {"smooth_window": 10, 
                    "general_threshold": 0.5,
                    "general_min_duration": 0.5, 
                    "rest_min_duration": 4,
                    "pre_locomotion_buffer": 5, 
                    "post_locomotion_buffer": 5, 
                    "locomotion_duration": 2}
```

```python
track2p_datas_paths = [os.path.join(track2p_datas_dir, f) for f in os.listdir(track2p_datas_dir) if os.path.isdir(os.path.join(track2p_datas_dir, f))]
for track2p_data_path in track2p_datas_paths:
    cell_type = track2p_data_path.split(os.sep)[-1]
    data_save_full_path = data_save_dir / f"{cell_type}.pickle"
    print(f"cell_type: {cell_type}")
    if not os.path.exists(data_save_full_path):
        print(f"Calculating data for {cell_type}...")
        raw_dffs, raw_intensitys = calculate_t2p_dff(track2p_data_path)
        groups_files = [os.path.join(groups_files_dir, f) for f in os.listdir(groups_files_dir) if f.endswith(".csv")]
        for groups_file in groups_files:
            if cell_type == os.path.basename(groups_file).split(".")[0]:
                print(f"group-cell_type: {os.path.basename(groups_file).split('.')[0]}")
                print(f"groups_file: {groups_file}")
                intensitys, dffs, runnings, imaging_frame_rates, running_sample_rates, events = load_and_align_data(raw_intensitys, raw_dffs, groups_file)
                save_dict = {
                            'intensitys': intensitys,
                            'dffs': dffs,
                            'runnings': runnings,
                            'imaging_frame_rates': imaging_frame_rates,
                            'running_sample_rates': running_sample_rates,
                            'events': events
                            }
                with open(data_save_full_path, 'wb') as f:
                    pickle.dump(save_dict, f)
                    
    else:
        print(f"Loading pre-calculated data for {cell_type}...")
        with open(data_save_full_path, 'rb') as f:
            loaded = pickle.load(f)
        intensitys = loaded['intensitys']
        dffs = loaded['dffs']
        runnings = loaded['runnings']
        imaging_frame_rates = loaded['imaging_frame_rates']
        running_sample_rates = loaded['running_sample_rates']
        events = loaded['events']
        
    # plot_overall_dff_and_running(dffs, runnings, imaging_frame_rates, running_sample_rates, events, plot_pre, plot_post, **running_parameters)
    sound_dffs, sound_runnings, sound_ensemble_sizes = calculate_event_dff_from_dff(dffs, runnings, imaging_frame_rates, running_sample_rates, events, time_window, pre_duration, post_duration, plot_pre, plot_post, **running_parameters)
    sound_dffs, sound_runnings = calculate_event_dff_from_intensity(intensitys, runnings, imaging_frame_rates, running_sample_rates, events, plot_pre, plot_post, **running_parameters)
    plot_event_dff(sound_dffs, imaging_frame_rates, plot_pre, cell_type, "#09af25", None, fig_save_path, fig_save_format)
    plot_event_running(sound_runnings, running_sample_rates, plot_pre, cell_type, "#000000", (-1, 1), fig_save_path, fig_save_format)
    event_dff_analysis(sound_dffs, plot_pre, plot_post, imaging_frame_rates, cell_type, fig_save_path, fig_save_format)
    event_running_analysis(sound_runnings, plot_pre, plot_post, running_sample_rates, cell_type, fig_save_path, fig_save_format)
```
